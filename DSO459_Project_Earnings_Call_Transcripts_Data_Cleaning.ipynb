```python
import pdfplumber
import os
```

```python
# Extracting unstructured text data and organizing it into a structured format for data analysis

folder_path = "/content/drive/MyDrive/DSO459/459 Earnings Calls Transcripts"
output_folder = "/content/drive/MyDrive/DSO459/459 Earnings Calls Transcripts/extracted_text"
os.makedirs(output_folder, exist_ok=True)

failed = []
success = 0

for filename in os.listdir(folder_path):
    if filename.endswith(".pdf"):
        pdf_path = os.path.join(folder_path, filename)
        txt_filename = filename.replace(".pdf", ".txt")
        txt_path = os.path.join(output_folder, txt_filename)

        try:
            with pdfplumber.open(pdf_path) as pdf:
                full_text = ""
                for page in pdf.pages:
                    text = page.extract_text()
                    if text:
                        full_text += text + "\n"

            with open(txt_path, "w", encoding="utf-8") as f:
                f.write(full_text)

            success += 1
            print(f"✓ {filename}")

        except Exception as e:
            failed.append(filename)
            print(f"✗ Failed: {filename} — {e}")

print(f"\nDone! Successfully converted: {success}")
print(f"Failed: {len(failed)}")
```

```python
#  Clean the top section (boilerplate) and only save the actual transcrip starting from "Presentation"
import re
import pandas as pd

input_folder = "/content/drive/MyDrive/DSO459/459 Earnings Calls Transcripts/extracted_text"
output_folder = "/content/drive/MyDrive/DSO459/459 Earnings Calls Transcripts/cleaned_text_v2"
os.makedirs(output_folder, exist_ok=True)
```

```python
# GenAI keyword list
keywords = [
    "generative ai", "generative artificial intelligence",
    "genai", "large language model", "llm", "gpt",
    "chatgpt", "foundation model", "diffusion model",
    "artificial intelligence", "machine learning", "deep learning", r"\bai\b"
]
# r"\bai\b" \b to implement word boundary so findings match "AI" but not words that contain "ai" like 'said'

results = []

for filename in os.listdir(input_folder):
    if filename.endswith(".txt"):
        filepath = os.path.join(input_folder, filename)

        with open(filepath, "r", encoding="utf-8") as f:
            text = f.read()

        # Cut to first occurrence of "Operator"
        occurrences = [m.start() for m in re.finditer(r'Operator', text)]
        if occurrences:
            text = text[occurrences[0]:]

        # Remove copyright lines
        text = re.sub(r'COPYRIGHT.*?\n', '', text)
        text = re.sub(r'spglobal\.com.*?\n', '', text)

        # Lowercase for counting
        text_lower = text.lower()
        total_words = len(text_lower.split())

        # Count keywords
        keyword_counts = {}
        for kw in keywords:
            if kw.startswith(r"\b"):
                count = len(re.findall(kw, text_lower))
            else:
                count = text_lower.count(kw)
            keyword_counts[kw] = count

        total_ai = sum(keyword_counts.values())
        normalized_score = total_ai / total_words if total_words > 0 else 0

        # Save cleaned file
        cleaned_path = os.path.join(output_folder, filename)
        with open(cleaned_path, "w", encoding="utf-8") as f:
            f.write(text)

        results.append({
            "filename": filename,
            "total_words": total_words,
            "total_ai_mentions": total_ai,
            "normalized_score": round(normalized_score, 6),
            **keyword_counts
        })

        print(f"✓ {filename} | words: {total_words} | AI: {total_ai} | Score: {normalized_score:.6f}")
```

```python
# Creating df and converting to csv
pd.set_option('display.max_columns', None)
df = pd.DataFrame(results)
df.to_csv("/content/drive/MyDrive/DSO459/459 Earnings Calls Transcripts/ai_scores.csv", index=False)
print(f"Processed {len(results)} transcripts")
print("\nTop 10 highest AI disclosure scores:")
print(df[["filename", "total_ai_mentions", "normalized_score"]].sort_values("normalized_score", ascending=False).head(10))
```

```python
# Checking df
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/DSO459/459 Earnings Calls Transcripts/ai_scores.csv")
print(df.columns.tolist())
print(df.head(2))
```

```python
# Parsing the filename to extract company name, quarter, and year so we can merge with financial data later
def parse_filename(filename):
    # ref: NVIDIA Corporation, Q3 2024 Earnings Call, Nov 21, 2023.txt
    # ref 2: T-Mobile US Inc., Q3 2025 Earnings Call, Oct 23, 2025.txt
    try:
        # Remove .txt
        name = filename.replace(".txt", "")
        # Split by comma
        parts = name.split(",")

        # Company name is everything before the first "Q\d \d{4}" match
        # Search all parts for the quarter pattern
        quarter = None
        year = None
        company_parts = []

        for part in parts:
            quarter_match = re.search(r'Q(\d)\s(\d{4})', part.strip())
            if quarter_match and quarter is None:
                quarter = f"Q{quarter_match.group(1)}"
                year = int(quarter_match.group(2))
                break  # everything before is company name
            else:
                company_parts.append(part.strip())

        company = ", ".join(company_parts)

        return company, quarter, year

    except Exception as e:
        return None, None, None

# loop through each filename
companies = []
quarters = []
years = []

for filename in df["filename"]:
    company, quarter, year = parse_filename(filename)
    companies.append(company)
    quarters.append(quarter)
    years.append(year)

df["company"] = companies
df["quarter"] = quarters
df["year"] = years
# Create a quarter_year column for easy reference
df["quarter_year"] = df["quarter"] + "_" + df["year"].astype(str)

# Check results
print(f"Null quarters: {df['quarter'].isnull().sum()}")
print(f"Null years: {df['year'].isnull().sum()}")
print(f"Unique companies: {df['company'].nunique()}")
print(f"\nSample:")
print(df[["company", "quarter", "year"]].head(10))

# Save updated CSV
df.to_csv("/content/drive/MyDrive/DSO459/459 Earnings Calls Transcripts/ai_scores.csv", index=False)
print("\nSaved parsed results")
```
